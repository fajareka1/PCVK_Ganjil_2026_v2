```python
# Ganti path sesuai file Anda
img = cv.imread("Untitled33_20260115083132.png")
imgRGB = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# ===== 1) figsize vs ukuran pixel =====
plt.figure(figsize=(3,3)); plt.imshow(imgRGB)
plt.title(f"figsize kecil - shape={imgRGB.shape}"); plt.show()

plt.figure(figsize=(10,10)); plt.imshow(imgRGB)
plt.title(f"figsize besar - shape={imgRGB.shape}"); plt.show()
# Kesimpulan: figsize hanya mengubah ukuran TAMPILAN plot di layar,
# img.shape (jumlah pixel asli) TIDAK berubah.
```

```python
# ===== 2) channel Red-Blue dan Green-Blue saja =====
img_RB = imgRGB.copy(); img_RB[:,:,1] = 0   # matikan Green
img_GB = imgRGB.copy(); img_GB[:,:,0] = 0   # matikan Red

plt.subplot(1,2,1); plt.imshow(img_RB); plt.title("Red-Blue")
plt.subplot(1,2,2); plt.imshow(img_GB); plt.title("Green-Blue")
plt.show()
```

```python
# ===== 3) crop baris 20-115, kolom 25-120 =====
img_crop = imgRGB[20:115, 25:120]
plt.imshow(img_crop); plt.title(f"shape={img_crop.shape}"); plt.show()

```

```python
# ===== 4) baris 5-30, semua kolom, channel Red saja =====
img_baris_R = imgRGB[5:30, :, 0]
plt.imshow(img_baris_R, cmap="Reds"); plt.title("Baris 5-30, channel Red"); plt.show()
```

```python
# ===== 5) 5 kotak ukuran & warna berbeda (random) =====
img_kotak5 = imgRGB.copy()
np.random.seed(1)
tinggi, lebar = img_kotak5.shape[0], img_kotak5.shape[1]
for i in range(5):
    x1 = np.random.randint(0, lebar-50)
    y1 = np.random.randint(0, tinggi-50)
    ukuran = np.random.randint(20, 80)
    warna = tuple(int(c) for c in np.random.randint(0, 255, 3))
    cv.rectangle(img_kotak5, (x1,y1), (x1+ukuran, y1+ukuran), warna, thickness=-1)
plt.imshow(img_kotak5); plt.title("5 kotak acak"); plt.show()
```

```python
# ===== 6) image posisi terbalik =====
img_flip = cv.flip(imgRGB, -1)   # -1 = terbalik horizontal + vertikal sekaligus
plt.imshow(img_flip); plt.title("Image terbalik"); plt.show()
```

```python
# ===== 7) rectangle & circle pada wajah (foto beraktivitas, bukan pasfoto) =====
img_aktivitas = cv.imread("/content/drive/MyDrive/VIaloga-vs-Teo.jpg")
img_aktivitas_rgb = cv.cvtColor(img_aktivitas, cv.COLOR_BGR2RGB)
# Sesuaikan koordinat berikut dengan posisi wajah pada foto Anda sendiri
cv.rectangle(img_aktivitas_rgb, (1270,500), (1420,360), (0,255,0), thickness=4)
cv.circle(img_aktivitas_rgb, (1345,430), 70, (255,0,0), thickness=4)
plt.imshow(img_aktivitas_rgb); plt.title("Rectangle & Circle pada wajah"); plt.show()
```

```python
# ===== 8 & 9) rectangle sudut bawah kiri channel B + tulisan nama file =====
nama_file = "VIaloga-vs-Teo.jpg"   # ganti sesuai citra pilihan: lena/mandrill/male/female/couple/sailboat/peppers
img_c = cv.imread(f"/content/drive/MyDrive/{nama_file}")
imgRGB_c = cv.cvtColor(img_c, cv.COLOR_BGR2RGB)
tinggi, lebar = imgRGB_c.shape[0], imgRGB_c.shape[1]

hasil = imgRGB_c.copy()
ukuran_kotak = 600
cv.rectangle(hasil, (0, tinggi-ukuran_kotak), (ukuran_kotak, tinggi), (0,255,0), thickness=4)

font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(hasil, nama_file, org=(10, tinggi-10), fontFace=font,
           fontScale=3.6, color=(255,255,0), thickness=2, lineType=cv.LINE_AA)

plt.imshow(hasil); plt.title("Rectangle sudut bawah kiri + nama file"); plt.show()

# Jika diminta khusus channel B saja (grayscale channel biru):
channel_B_only = imgRGB_c[:,:,2]
plt.imshow(channel_B_only, cmap="Blues"); plt.title("Channel B saja"); plt.show()
```
